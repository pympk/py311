```python
# Add this to a temporary script or run in a notebook
import pandas as pd
from core.paths import LOCAL_DATA_DIR

# df_ohlcv = pd.read_parquet(GLOBAL_PROCESSED_DIR / "df_ohlcv.parquet")
df_ohlcv = pd.read_parquet(LOCAL_DATA_DIR / "df_ohlcv.parquet")
features_df = pd.read_parquet(LOCAL_DATA_DIR / "features_df.parquet")
alpha_df = pd.read_parquet(LOCAL_DATA_DIR / "alpha_cache_41d_1998.parquet")

ohlcv_tickers = set(df_ohlcv.index.get_level_values("Ticker").unique())
feat_tickers = set(features_df.index.get_level_values("Ticker").unique())
alpha_tickers = set(alpha_df.index.get_level_values("Ticker").unique())

len_ohlcv = len(ohlcv_tickers)
len_feat = len(feat_tickers)
len_alpha = len(alpha_tickers)
```

```python
import numpy as np

my_item = "SPY"
# my_item = "DVN"
decision_date = "2022-04-01"


df = df_ohlcv.loc[my_item].loc[decision_date:].head(7).copy()
log_return = np.log(df["Adj Close"].iloc[-1] / df["Adj Close"].iloc[1])
mkt_return = np.exp(log_return) - 1
print(my_item)
print(df)
print(f"\ndf_log_return: {log_return:.6f}")
print(f"mkt_return: {mkt_return:.6f}")
```

```python
raw_log_reward = -0.038901
slippage_applied = 0.001
simple_reward = np.exp(raw_log_reward) - 1
actual_return = simple_reward - slippage_applied

print(f"raw_log_reward: {raw_log_reward:.6f}")
print(f"simple_reward:  {simple_reward:.6f}")
print(f"actual_return:  {actual_return:.6f}")
```

```python
alpha = actual_return - mkt_return
print(f"alpha:  {alpha:.6f}")
```

```python
alpha = -0.002

downside_penalty: float = 1.5  # 1x penalty on underperformance (Alpha < 0)

# Penalize underperformance aggressively for the RL Agent
penalized_alpha = alpha * downside_penalty if alpha < 0 else alpha
print(f"penalized_alpha: {penalized_alpha:.6f}")
```

```python
holding_period = 5

# Math scales it down by holding period logic per capital deployment
portfolio_impact = actual_return / holding_period
print(f"portfolio_impact: {portfolio_impact:.6f}")
```

```python
alpha_impact = alpha / holding_period
print(f"alpha_impact: {alpha_impact:.6f}")
```

```python
equity_curve = [1.0]

equity_curve.append(equity_curve[-1] * (1.0 + portfolio_impact))
agent_equity = equity_curve[-1]
print(f"agent_equity: {agent_equity:.6f}")
```

```python
alpha_equity_curve = [1.0]

alpha_equity_curve.append(alpha_equity_curve[-1] * (1.0 + alpha_impact))
alpha_equity = alpha_equity_curve[-1]
print(f"alpha_equity: {alpha_equity:.6f}")
```
